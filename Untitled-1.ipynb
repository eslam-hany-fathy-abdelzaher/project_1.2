```python
import time
import tracemalloc
import pandas as pd
import polars as pl
import numpy as np

# ==========================================
# 1. Data Generation (توليد بيانات كبيرة للتجربة)
# ==========================================
# ملاحظة: إذا كان لديك ملف جاهز مثل car_prices.csv، قم بوضع مساره في المتغير DATA_FILE
DATA_FILE = "dummy_sales_data.csv"

def generate_dummy_data(filename, num_rows=2_000_000):
    print(f"Generating {num_rows:,} rows of sales data...")
    np.random.seed(42)
    categories = ['Electronics', 'Clothing', 'Home', 'Books', 'Toys']
    
    dates = pd.date_range(start='2023-01-01', end='2023-12-31', periods=1000).astype(str)
    
    data = {
        'transaction_id': range(1, num_rows + 1),
        'category': np.random.choice(categories, size=num_rows),
        'amount': np.random.uniform(10.0, 500.0, size=num_rows),
        'timestamp': np.random.choice(dates, size=num_rows),
    }
    
    df = pd.DataFrame(data)
    # إضافة بضع قيم مفقودة للفلترة
    df.loc[df.sample(frac=0.01).index, 'amount'] = None 
    df.to_csv(filename, index=False)
    print("Data generation complete!\n")

# قم بتوليد البيانات أول مرة فقط
generate_dummy_data(DATA_FILE, num_rows=2_000_000)


# ==========================================
# 2. Pandas Pipeline
# ==========================================
def run_pandas(file_path):
    tracemalloc.start()
    start_time = time.time()
    
    # Ingestion & Cleaning
    df = pd.read_csv(file_path)
    df = df.dropna()  # Filter missing values
    
    # Datetime Parsing
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    df['month'] = df['timestamp'].dt.to_period('M')
    
    # Complex Grouping & Sorting
    result = df.groupby(['category', 'month'])['amount'].sum().reset_index()
    result = result.sort_values(by=['month', 'amount'], ascending=[True, False])
    
    end_time = time.time()
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    execution_time = end_time - start_time
    peak_memory_mb = peak_memory / (1024 * 1024)
    
    return execution_time, peak_memory_mb


# ==========================================
# 3. Polars Pipeline
# ==========================================
# ==========================================
# 3. Polars Pipeline (المعدلة)
# ==========================================
def run_polars(file_path):
    tracemalloc.start()
    start_time = time.time()
    
    # Ingestion & Automatic Datetime Parsing
    df = pl.read_csv(file_path, try_parse_dates=True)
    
    result = (
        df.filter(pl.col("amount").is_not_null())  # Filter missing values
        .with_columns(
            pl.col("timestamp")
            .cast(pl.String)
            .str.to_datetime(strict=False)
            .dt.truncate("1mo")
            .alias("month") # Parse date safely
        )
        .group_by(["category", "month"])
        .agg(pl.col("amount").sum().alias("total_revenue")) # Grouping
        .sort(["month", "total_revenue"], descending=[False, True]) # Sorting
    )
    
    end_time = time.time()
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    execution_time = end_time - start_time
    peak_memory_mb = peak_memory / (1024 * 1024)
    
    return execution_time, peak_memory_mb
# ==========================================
# 4. Performance Profiling & Report Execution
# ==========================================
print("Running Pandas Pipeline...")
pandas_time, pandas_mem = run_pandas(DATA_FILE)

print("Running Polars Pipeline...")
polars_time, polars_mem = run_polars(DATA_FILE)

# حساب فروقات الأداء
time_diff_pct = ((pandas_time - polars_time) / pandas_time) * 100
mem_diff_pct = ((pandas_mem - polars_mem) / pandas_mem) * 100

# ==========================================
# 5. Markdown Report Generation
# ==========================================
print("\n" + "="*50)
print("FINAL PERFORMANCE REPORT")
print("="*50 + "\n")

report_markdown = f"""
| Metric | Pandas Pipeline | Polars Pipeline | % Performance Difference |
| :--- | :--- | :--- | :--- |
| **Execution Speed** | {pandas_time:.2f} seconds | {polars_time:.2f} seconds | Polars is {time_diff_pct:.1f}% faster |
| **Peak Memory Used** | {pandas_mem:.2f} MB | {polars_mem:.2f} MB | Polars saved {mem_diff_pct:.1f}% RAM |
"""

print(report_markdown)
```
